##### break

```python
'''
當偵測到字母 t 時，就會強制結束迴圈
'''
for char in "contents":
    print(char, end=" ")
    if char == "t":
        break

print("\n迴圈結束")
```

##### continue

```python
'''
當偵測到字母 t 時，
會跳過本次迴圈剩下的程式碼 print(char)，
但不會結束迴圈，仍然會進入下一圈繼續執行
'''
for char in "contents":
    if char == "t":
        continue
    print(char, end=" ")

print("\n迴圈結束")
```

### 小試身手
計算 1 到 100 之間的和，如果值超過500就停止執行

```python
limit_value = 500
total = 0

for i in range(1, 101):
    if total > limit_value:
        print("\n",i - 1 , total)
        break
    else:
        total += i
        print(i, end=", ")
```

計算 1 到 100 之間的奇數之和，但要跳過所有被 5 整除的數

```python
total = 0
for i in range(1, 101):
    if i % 5 == 0:
        continue
    elif i % 2 != 0: 
        total += i

print(total)

# Teacher
total = 0
for i in range(1, 101, 2):
    if i % 5 == 0:
        continue
    total += i

print(total)
```
