| 用法 | 說明 |
|:---:|:---:|
| `range(m)` | 走訪 0 到 m-1 的數字 |
| `range(n, m)` | 走訪 n 到 m-1 的數字  |
| `range(n, m, step)` | 以step為間距，走訪 n 到 m-1 的數字 |

##### for

```python
# for <variable> in <可迭代物件>:
for char in "contents":
    print(char, end=", ")
```

```python
# range(m-1)
for i in range(5):
    print(i)
```

```python
# range(n, m-1)
for i in range(1, 11):
    print(i, end=", ")
```

```python
# range(n, m-1, step)
for i in range(1, 11, 2):
    print(i, end=" ")
```

### 小試身手 for  
請產出 a 開頭的 n 個偶數(含 a 本身)  
Ex:  
a = 5, n = 3  
Output : 6 8 10

a = 4, n = 5  
Output : 4 6 8 10 12

```python
a = 4
n = 5

# first try
# if a % 2 != 0:
#     a += 1
# maximun = a + 2 * (n - 1) + 1

# for i in range(a, maximun, 2):
#     print(i, end=" ")

for i in range(n):
    if a % 2 == 0:
        print(a + i * 2, end=" ")
    else:
        print(a + 1 + i * 2, end=" ")
```

```python
# 嘗試產出
"""
* * * * *
 * * * *
  * * *
   * *
    *
"""
# 1
num = 5
result = ""
for i in range(num):
    result += " " * i + "* " * (num - i - 1) + "*" + " " * i + "\n"
print(result)

# 2
num = 5
for i in range(num):
    print(" " * i + "* " * (num - i))

print()
# 嘗試產出
"""
    *
   * *
  * * *
 * * * *
* * * * *
"""
# 1
num = 5 
result = ""
for i in range(num):
    result += " " * (num - i - 1) + "* " * i + "*" + " " * (num - i - 1) + "\n"
print(result)

# 2
num = 5
for i in range(num):
    print(" " * (num - i - 1) + "* " * (i + 1))
```

##### while

```python
count = 1

while count <= 5:
    print(count, end=" ")
    count += 1

print("\ncount 已經不小於 5")
print(count)
```
