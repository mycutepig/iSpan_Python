條件語句 if-elif-else

```python
'''
條件判斷
if ...
if ... else ...
if ... elif ... else ...

'''
```

##### if

```python
num = 20
if num > 10:
    print("Num 大於 10")
```

##### if...else...

```python
num = 2
if num > 10:
    print("Num 大於 10")
else:
    print("Num 沒有大於 10")
```

##### if...elif... (else)

```python
num = 5.5
if num > 10:
    print("Num 大於 10")
elif num % 2 == 0:
    print("Num 是偶數")
elif type(num) == int:
    print("Num 資料型態為 int")
else:
    print("Num 沒有大於 10 且不為 int")
```
