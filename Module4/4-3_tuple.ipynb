```python
# tuple 初始化: 第一種
data = ("Big", "Data", "is", "Good")
print(data)
print(type(data))
```

```python
# tuple 初始化: 第二種
data = "Piggy", 18
print(data)
print(type(data))

x, y = data
print(x)
print(type(x))
print(y)
print(type(y))

name, age = "Piggy", 18
print(type(name))
print(type(age))
```

```python
# 透過指定索引輸出值
data = "Piggy", 18, "AIPE02"
print(data)
print(data[0])
```

```python
# 複數變數修改值

```

```python
# tuple 不可以修改指定索引的值
data[0] = "Jeff"
# tuple 不可修改
```

```python
# ( list )可以修改指定索引的值
num_list = [1, 2, 3]
num_list[0] = 100
print(num_list)
```

```python
# 用 for 迴圈逐一輸出資料
data = "Piggy", 18, "AIPE02"
for i in data:
    print(i)
```

```python
# 用 len 計算 tuple 資料個數
data = "Piggy", 18, "AIPE02"
print(data)
print(len(data))
print(data[0])
```
