```python
# 初始化一個 list
ids = [1, 2, 3, 4, 5, 6]
print(ids)
```

```python
# 新增元素在 list 尾端
# list.append(<value>)
ids = [1, 2, 3, 4, 5, 6]
ids.append(7)
print(ids)
```

```python
ids = [1, 2, 3, 4, 5, 6]
print(ids[3])
```

```python
# 修改索引位置的元素
ids = [1, 2, 3, 4, 5, 6]
ids[3] = 99
print(ids)
```

```python
# 刪除索引位置的元素
# list.pop(<index>)
ids = [1, 2, 3, 4, 5, 6]
ids.pop(3)
print(ids)
```

```python
# 在指定索引 1, 新增元素 9 ，其餘元素往後移
# list.insert(<index>, <value>)
ids = [1, 2, 3, 4, 5, 6]
ids.insert(1, 9)
print(ids)
```

```python
# 移除指定的值
# list.remove(<value>)
ids = [1, 2, 3, 4, 5, 6]
ids.remove(4)
print(ids)
```

```python
ids = [1, 2, 3, 4, 5, 6]
for _ in range(10):
    ids.insert(1, 9)
print(ids)
```

```python
print(ids)

# 錯誤,不可以在 <list> for loop 裡修改 <list>, index 可能會錯
# for num in ids:
#     if num == 9:
#         ids.remove(9)

while 9 in ids:
    ids.remove(9)

print(ids)
```

```python
ids = [1, 2, 3, 4, 5, 6]
for _ in range(5):
    ids.append(10)
print(ids)

# 計算 list 內的某個值 共有幾個
# list.count(<value>)
cnt_10 = ids.count(10)
print(cnt_10)
```

```python
# list 的合併
ids1 = [1, 2, 3, 4]
ids2 = [5, 6, 7]

for num in ids2:
    ids1.append(num)
print(ids1)
```

```python
# list 的合併
# list.extend(<list>)
ids1 = [1, 2, 3, 4]
ids2 = [5, 6, 7]
ids1.extend(ids2)
print(ids1)
```

```python
ids1 = [1, 2, 3, 4]
ids2 = [5, 6, 7]

total_ids = ids1 + ids2
print(total_ids)
```

### list 搭配 for 迴圈

```python
# 建立一個 1~10 的 list
ls = []
for i in range(1, 11):
    ls.append(i)
print(ls)
```

```python
# 建立一個 1~50 的 list, 且只要偶數
ls = []
start = 1
end = 50
if start % 2 != 0:
    start += 1
for i in range(start, end + 1, 2):
    ls.append(i)
print(ls)
```

```python
# 建立一個 list, 裡面要放遞增的 list
inside_ls = []
result = []
for num in range(1, 11):
    inside_ls.append(num)
    result.append(inside_ls[:])
print(result)

# outside = []
# for i in range(1, 11):
#     inside = []
#     for c in range(1, i + 1):
#         inside.append(c)
#     outside.append(inside)
# print(outside)
```

```python
# enumerate() 同時取得 index 與 value
ids = [10, 2, 35, 4, 15]
for index, value in enumerate(ids):
    print(f"index: {index}, value: {value}")
```

### List comprehension 列表推導式
`new_list = [expression for item in iterable [if condition]]`  
  
`list的名稱 = [表達式 for 元素 in 迭代對象 [if 判斷式(...的條件)]]`

```python
# 建立一個 1~10 的 list
ls = [i for i in range(1, 11)]
print(ls)

# 建立一個 1~50 的 list, 且只要偶數
ls = [i for i in range(2, 51, 2)]
print(ls)
# or 
ls = [i for i in range(1, 51) if i % 2 == 0]
print(ls)

# 建立一個 list, 裡面要放遞增的 list
outside = []
for i in range(1, 11):
    outside.append([c for c in range(1, i + 1)])
print(outside)
# 再精簡
outside = [[c for c in range(1, i + 1)] for i in range(1, i + 1)]
print(outside)
```

```python
# 1~10 的二次方
ls = [i ** 2 for i in range(1, 11)]
print(ls)
```
