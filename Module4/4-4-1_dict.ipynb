```python
'''
13. Roman to Integer
Roman numerals are represented by seven different symbols: I, V, X, L, C, D and M.

Symbol       Value
I             1
V             5
X             10
L             50
C             100
D             500
M             1000
For example, 2 is written as II in Roman numeral, just two ones added together. 12 is written as XII, which is simply X + II. The number 27 is written as XXVII, which is XX + V + II.

Roman numerals are usually written largest to smallest from left to right. However, the numeral for four is not IIII. Instead, the number four is written as IV. Because the one is before the five we subtract it making four. The same principle applies to the number nine, which is written as IX. There are six instances where subtraction is used:

I can be placed before V (5) and X (10) to make 4 and 9. 
X can be placed before L (50) and C (100) to make 40 and 90. 
C can be placed before D (500) and M (1000) to make 400 and 900.
Given a roman numeral, convert it to an integer.
'''
def roman_to_integer(s):
    dict1 = {
        "I": 1,
        "V": 5,
        "X": 10,
        "L": 50,
        "C": 100,
        "D": 500,
        "M": 1000
    }
    dict2 = {
        "IV": 4,
        "IX": 9,
        "XL": 40,
        "XC": 90,
        "CD": 400,
        "CM": 900
    }
    result = 0
    for symbol in dict2:
        while symbol in s:
            result += dict2[symbol]
            s = s.replace(symbol, "")

    for symbol in dict1:
        while symbol in s:
            result += dict1[symbol]
            s = s.replace(symbol, "", 1)
    return result

print(roman_to_integer("MCMXCIV"))
```

##### dict 的合併

```python
# dict 的合併
# dict.update()
fruit_dict2 = {
    "kiwi": {
        "price": 30,
        "amount": 20,
        "color": ["green"]
    }
}

fruit_dict.update(fruit_dict2)
print(fruit_dict)
```

```python
# 補充 合併 or 修改 dict 的方法
print(fruit_dict)
fruit_dict3 = {
    **fruit_dict,
    "kiwi": {
        "price": 100,
        "amount": 50,
        "color": ["yellow"]
    }
}
print(fruit_dict3)

# or 
fruit_dict = {
    **fruit_dict,
    "kiwi": {
        "price": 100,
        "amount": 50,
        "color": ["yellow"]
    }
}
print(fruit_dict)
```

##### 另外的 print 方式 `pprint` (pretty-print) 

```python
# 從 Module 引入 function
from pprint import pprint
print(fruit_dict3)
# pprint 打印出來的資料 沒有按照順序
pprint(fruit_dict3)
```

```python
'''
新增顏色
apple 有新的品種 顏色要有黃色
tomato 有綠色
'''
fruit_dict3["apple"]["color"].append("yellow")
fruit_dict3["tomato"]["color"].append("green")

pprint(fruit_dict3)
```
