```python
# f-string (又稱作 formatted string literals, version >= 3.6)
name = "Pig"
age = 18

print(f"{name} is {age} years old.")
```

```python
# 靠左對齊、靠右對齊
msg = f"name: [{name:<10}], age: [{age:>10}]"
print(msg)
```

```python
# 靠左對齊、靠右對齊，字數不足，補「ㄏ」
# (沒寫補什麼，預設空格)
msg = f"name: [{name:x<10}], age: [{age:x>10}]"
print(msg)
```

```python
# 置中對齊
msg = f"name: [{name:^10}], age: [{age:^6}]"
print(msg)
```

```python
# 指定文字長度上限 (只有文字，才在格式化字串中加入「.」來限定字串長度)
msg = f"name: [{name:.2}]"
print(msg)
```

```python
data = 3.1415926
print(f"{data:.2f}")
```
