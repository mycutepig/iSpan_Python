```python
fruit_dict = {    
    'apple': {
        'price': 100,
        'quantity': 50,
        'color': ['greem', 'red']
    },
    'orange': {
        'price': 50,
        'quantity': 30,
        'color': 'orange' # 錯誤的寫法
    }
}
print(fruit_dict)
```

```python
# 使用 for 迴圈讀取 dict 資料1 
# 只會給 key 值
for key in fruit_dict:
    print(key)
```

```python
# 使用 for 迴圈讀取 dict 資料2
# 取得 dict 的 value
# dict.values()
for value in fruit_dict.values():
    print(value)
```

```python
# 使用 for 迴圈讀取 dict 資料3
# 取得 dict 的 key, value
# dict.items()
for key, value in fruit_dict.items():
    print(key)
    print(value)
```

```python
# 取得 dict 的 value
fruit_dict = {    
    'apple': {
        'price': 100,
        'quantity': 50,
        'color': ['greem', 'red']
    },
    'orange': {
        'price': 50,
        'quantity': 30,
        'color': 'orange' # 錯誤的寫法
    }
}
print(fruit_dict)

# print(fruit_dict["kiwi"]) 導致 KeyError
print(fruit_dict.get("apple"))
print(fruit_dict.get("kiwi"))
# 沒有該 key 回傳 None, 不會導致程式錯誤
```

```python
# 資料處理
fruit_color_list = [
    {'apple': 'red'}, 
    {'grape': 'purple'}, 
    {'kiwi': 'yellow'}, 
    {'apple': 'green'}, 
    {'kiwi': 'green'}, 
    {'grape': 'green'}
]
```

```python
# 將資料整理成 {'apple': ['red', 'green'], 'grape': ['purple', 'green'], 'kiwi': ['yellow', 'green']}
fruit_color_dict = {}
for data in fruit_color_list:
    for fruit, color in data.items():
        # 如果沒有該 key 值就執行
        if fruit_color_dict.get(fruit) == None:
            fruit_color_dict[fruit] = [color]
        # 如果有該 key 值就執行
        else:
            fruit_color_dict[fruit].append(color)

print(f"最終的 color_data: {fruit_color_dict}")
```

##### `defaultdict(<type>)`

```python
from collections import defaultdict
# 當訪問的鍵不存在時，可以自動創建 key 和 預設值 ，而不是拋出KeyError。
data = defaultdict(int)

# 以 int 型態的預設值 0 來初始化
print(data["price"])
print(data)
```

```python
# 資料處理
fruit_color_list = [
    {'apple': 'red'}, 
    {'grape': 'purple'}, 
    {'kiwi': 'yellow'}, 
    {'apple': 'green'}, 
    {'kiwi': 'green'}, 
    {'grape': 'green'}
]
```

```python
# 將資料整理成 {'apple': ['red', 'green'], 'grape': ['purple', 'green'], 'kiwi': ['yellow', 'green']}

from collections import defaultdict
color_data = defaultdict(list)

for fruit_color in fruit_color_list:
    print(fruit_color)
    for name, color in fruit_color.items():
        color_data[name].append(color)
print(color_data)
```
