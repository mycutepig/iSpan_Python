##### 將 data.txt 的資料分類成另外三個檔案，分別為 cc.txt 、 css.txt 、 c.txt  

```python
with open("./data.txt", "r", encoding="utf-8") as f:
    doc_lists = f.readlines()

print(doc_lists)

list_c = []
list_cc = []
list_css = []
for data in doc_lists:
    filename_ext = data.split(".")[1].strip("\n")
    if filename_ext == "c":
        list_c.append(data)
    elif filename_ext == "cc":
        list_cc.append(data)
    elif filename_ext == "css":
        list_css.append(data)

with open("./c.txt", "w", encoding="utf-8") as f:
    f.write("".join(list_c))

with open("./cc.txt", "w", encoding="utf-8") as f:
    f.write("".join(list_cc))

with open("./css.txt", "w", encoding="utf-8") as f:
    f.write("".join(list_css))
```
