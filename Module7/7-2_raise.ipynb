```python
# 自訂例外處理
x = -1

if x < 0:
    raise Exception("資料不能小於 0")
```

```python
# 自訂型別錯誤(TypeError)的例外處理
msg = 100
if type(msg) != str:
    raise TypeError(f"資料必須為字串 使用者傳入了: {type(msg)}")
```

##### 設計 raise 陳述

```python
num = -3

try:
    if num <= 0:
        raise ValueError("輸入必須是正數！")
except ValueError as e:
    print(f"❌ 錯誤：{e}")
else:
    print("✅ 輸入正確！")
```
