```python
# 全域變數
name = "Pig"

# 定義函式
def greeting():
    msg = "Hello"
    name = "User"

    return msg + " " + name
# 執行函式
data = greeting()
print(data)

# 輸出 name
print(name)
```

##### 使用 global 讓函式存取全域變數的值

```python
# 全域變數
name = "Pig"

# 定義函式
def greeting():
    msg = "Hello"
    global name # 呼叫全域變數
    name = "User"

    return msg + " " + name
# 執行函式
data = greeting()
print(data)

# 輸出 name
print(name)
```
