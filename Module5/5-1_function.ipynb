```python
# 建立 function 3
def greeting(msg, name):
    str_list = msg.split()
    str_list.append(name)
    return str_list

data = greeting("Hello World!", "Pig")
print(data)
data = greeting(name="Pig", msg="Hello World!")
print(data)
```

```python
# 建立 function 4
# default parameter 一定要放在 def 的最後面
def greeting(name, msg="Good Morning"):
    str_list = msg.split()
    str_list.append(name)
    return str_list

data = greeting("Pig")
print(data)
data = greeting("Pig", "Hello World!")
print(data)
```

```python
# 改寫 Module2 practice
def get_list_odd_num(num_list):
    for num in num_list:
        if num % 2 == 0:
            num_list.remove(num)
    return num_list

print(get_list_odd_num([1, 2, 3, 4, 5, 6, 7]))
```

```python
def get_discount(price, discount):
    discount_index = price // 2000

    if price >= 2000:
        price -= discount * discount_index
    
    return price

print(get_discount(8800, 200))

data = [
    {
        "total": 6000,
        "discount": 200
    },
    {
        "total": 8000,
        "discount": 200
    }
]

for d in data:
    answer = get_discount(d["total"], d["discount"])
    print(f"折扣後的金額: {answer} 元")
```

```python
# args => tuple, kwargs => dict
def add_nums(*args, **kwargs):
    print(args)
    print(kwargs)

    sum = 0
    for i in args:
        sum += i
    print(sum)

    for key, value in kwargs.items():
        sum += value

    return sum
    

answer = add_nums(1, 2, 3, 4, 5, n1=100, n2=200)
print(answer)
```
