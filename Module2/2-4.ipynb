成員運算子 in / not in

```python
if 5 not in num_list:
    print("5 isn't in list!")
```

身分運算子 is / is not

```python
x = 10
y = 10

print(x)
print(y)
print(id(x))
print(id(y))
```

```python
if x is y:
    print("They share the same memory location!")
```

```python
if x == y:
    print("x is equal to y!")
```

```python
x = [1, 2, 3]
y = [1, 2, 3]

if x == y:
    print("x is equal to y!")

if x is y:
    print("They share the same memory location!")
print(id(x))
print(id(y))

y = x

if x is y:
    print("They share the same memory location!")
print(id(x))
print(id(y))
```

```python
parse_ok = True
if parse_ok is True:
    print("Correct!")

print(id(parse_ok))
print(id(True))
print(id(False))
```

位元運算子

```python
x = 1 # 0001
y = 3 # 0011

# Binary
print(x & y) # 0001
print(x | y) # 0011
print(x ^ y) # 0010
```
